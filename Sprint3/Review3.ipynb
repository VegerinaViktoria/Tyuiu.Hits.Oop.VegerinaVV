```polyglot-notebook
using System;
using System.Collections.Generic;

public interface IInstallable
{
    void Install();
}

public abstract class Component
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }
    private decimal _price;

    public decimal Price
    {
        get { return _price; }
        set
        {
            if (value >= 0)
                _price = value;
            else
                throw new ArgumentOutOfRangeException("Цена не может быть отрицательной!");
        }
    }

    protected Component(string model, string manufacturer, decimal price)
    {
        Model = model;
        Manufacturer = manufacturer;
        Price = price;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine($"Модель: {Model}, Производитель: {Manufacturer}, Цена: {Price} р.");
    }

    public override string ToString()
    {
        return $"{GetType().Name} | Модель: {Model}, Производитель: {Manufacturer}, Цена: {Price} р.";
    }
}

public class CPU : Component, IInstallable
{
    public int Cores { get; set; }
    public double ClockFrequencyGHz { get; set; }

    public CPU(string model, string manufacturer, decimal price, int cores, double clockFrequencyGHz)
     : base(model, manufacturer, price)
    {
        Cores = cores;
        ClockFrequencyGHz = clockFrequencyGHz;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"CPU | Модель: {Model}, Производитель: {Manufacturer}, Количество ядер: {Cores}, Тактовая частота: {ClockFrequencyGHz} ГГц, Цена: {Price} р.");
    }

    public void Install()
    {
        Console.WriteLine($"CPU {Model} установлен!");
    }

    public static bool operator ==(CPU a, CPU b)
    {
        if (ReferenceEquals(a, b))
            return true;
        if (a is null || b is null)
            return false;

        return a.Model == b.Model && a.Manufacturer == b.Manufacturer && a.Cores == b.Cores && a.ClockFrequencyGHz == b.ClockFrequencyGHz;
    }

    public static bool operator !=(CPU a, CPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is CPU other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Model, Manufacturer, Cores, ClockFrequencyGHz);
    }

}

public class RAM : Component, IInstallable
{
    public int MemoryCapacityGB { get; set; }
    public int FrequencyMHz { get; set; }

    public RAM(string model, string manufacturer, decimal price, int memoryCapacityGB, int frequencyMHz)
        : base(model, manufacturer, price)
    {
        MemoryCapacityGB = memoryCapacityGB;
        FrequencyMHz = frequencyMHz;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"RAM | Модель: {Model}, Производитель: {Manufacturer}, Объем памяти: {MemoryCapacityGB} ГБ, Тактовая частота: {FrequencyMHz} МГц, Цена: {Price} р.");
    }

    public void Install()
    {
        Console.WriteLine($"RAM {Model} установлена!");
    }

    public static bool operator ==(RAM a, RAM b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Model == b.Model && a.Manufacturer == b.Manufacturer && a.MemoryCapacityGB == b.MemoryCapacityGB && a.FrequencyMHz == b.FrequencyMHz;
    }

    public static bool operator !=(RAM a, RAM b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is RAM other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Model, Manufacturer, MemoryCapacityGB, FrequencyMHz);
    }
}

public class GPU : Component, IInstallable
{
    public int VideoMemoryGB { get; set; }
    public int MemoryBusWidth { get; set; }

    public GPU(string model, string manufacturer, decimal price, int videoMemoryGB, int memoryBusWidth)
        : base(model, manufacturer, price)
    {
        VideoMemoryGB = videoMemoryGB;
        MemoryBusWidth = memoryBusWidth;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"GPU | Модель: {Model}, Производитель: {Manufacturer}, Видеопамять: {VideoMemoryGB} ГБ, Разрядность шины видеопамяти: {MemoryBusWidth} бит, Цена: {Price} р.");
    }

    public void Install()
    {
        Console.WriteLine($"GPU {Model} установлена!");
    }

    public static bool operator ==(GPU a, GPU b)
    {
        if (ReferenceEquals(a, b))
            return true;
        if (a is null || b is null)
            return false;

        return a.Model == b.Model && a.Manufacturer == b.Manufacturer && a.VideoMemoryGB == b.VideoMemoryGB && a.MemoryBusWidth == b.MemoryBusWidth;
    }

    public static bool operator !=(GPU a, GPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is GPU other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Model, Manufacturer, VideoMemoryGB, MemoryBusWidth);
    }
}

public class HDD : Component, IInstallable
{
    public int CapacityGB { get; set; }
    public int SpeedRPM { get; set; }

    public HDD(string model, string manufacturer, decimal price, int capacityGB, int speedRPM)
        : base(model, manufacturer, price)
    {
        CapacityGB = capacityGB;
        SpeedRPM = speedRPM;
    }

    public override void GetInfo()
    {
        Console.WriteLine($"HDD | Модель: {Model}, Производитель: {Manufacturer}, Объем памяти: {CapacityGB} ГБ, Скорость: {SpeedRPM} об/мин, Цена: {Price} р.");
    }

    public void Install()
    {
        Console.WriteLine($"HDD {Model} установлен!");
    }

    public static bool operator ==(HDD a, HDD b)
    {
        if (ReferenceEquals(a, b))
            return true;
        if (a is null || b is null)
            return false;

        return a.Model == b.Model && a.Manufacturer == b.Manufacturer && a.CapacityGB == b.CapacityGB && a.SpeedRPM == b.SpeedRPM;
    }

    public static bool operator !=(HDD a, HDD b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is HDD other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Model, Manufacturer, CapacityGB, SpeedRPM);
    }
}

public class Computer
{
    private List<Component> _components = new List<Component>();

    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        _components.Add(component);
        component.Install();
    }

    public void RemoveComponent<T>(T component) where T : Component
    {
        if (_components.Remove(component))
            Console.WriteLine($"Компонент {component.Model} удален из компьютера");
        else
            Console.WriteLine($"Компонент {component.Model} не найден в компьютере");
    }

    public decimal GetTotalPrice()
    {
        decimal summa = 0;
        foreach (var component in _components)
            summa += component.Price;
        return summa;
    }

    public void ShowAllComponents()
    {
        Console.WriteLine("Компоненты компьютера:");
        foreach (var component in _components)
            component.GetInfo();
        
    }
}

public static class ComputerFactory
{
    public static Computer CreateOfficeComputer()
    {
        Computer computer = new Computer();
        computer.AddComponent(new CPU("Intel Core i3-12100", "Intel", 12000m, 4, 3.3));
        computer.AddComponent(new RAM("Kingston FURY Beast 16GB", "Kingston", 5000m, 16, 3200));
        computer.AddComponent(new HDD("WD Blue 1TB", "Western Digital", 4500m, 1000, 7200));
        return computer;
    }

    public static Computer CreateGameComputer()
    {
        Computer computer = new Computer();
        computer.AddComponent(new CPU("Intel Core i5-12400F", "Intel", 18000m, 6, 2.5));
        computer.AddComponent(new RAM("Kingston FURY Beast 32GB", "Kingston", 10000m, 32, 3200));
        computer.AddComponent(new GPU("NVIDIA GeForce RTX 4060", "NVIDIA", 45000m, 8, 128));
        computer.AddComponent(new HDD("WD Blue 2TB", "Western Digital", 7000m, 2000, 7200));
        return computer;
    }
    public static Computer CreateCustomComputer<T1, T2>(T1 component1, T2 component2) where T1 : Component,
     IInstallable where T2 : Component, IInstallable
    {
        Computer computer = new Computer();
        computer.AddComponent(component1);
        computer.AddComponent(component2);
        return computer;
    }
}

Console.WriteLine("--Основная программа--");
CPU cpu = new CPU("AMD Ryzen 7 7800X3D", "AMD", 45000m, 8, 4.2);
CPU cpu1 = new CPU("Intel Core i5", "Intel", 20000m, 6, 2.9);
CPU cpu2 = new CPU("Intel Core i5", "Intel", 20000m, 6, 2.9);
RAM ram = new RAM("Corsair Vengeance 16GB", "Corsair", 8000m, 16, 3600);
GPU gpu = new GPU("AMD Radeon RX 7900 XTX", "AMD", 110000m, 24, 384);
GPU gpu1 = new GPU("NVIDIA RTX 4090", "NVIDIA", 180000m, 24, 384);
GPU gpu2 = new GPU("AMD RX 7900 XTX", "AMD", 110000m, 24, 384);
HDD hdd = new HDD("Seagate Barracuda 4TB", "Seagate", 9000m, 4000, 7200);

Console.WriteLine("\n--Демонстрация полиморфизма--");
Component[] components = {cpu, ram, gpu, hdd};
foreach (var component in components)
{
    component.GetInfo();
}

Console.WriteLine("\n--Перегрузка операторов сравнения--");
Console.WriteLine($"cpu1: {cpu1}");
Console.WriteLine($"cpu2: {cpu2}");
Console.WriteLine($"cpu1 == cpu2: {cpu1 == cpu2}"); 
Console.WriteLine($"gpu1: {gpu1}");
Console.WriteLine($"gpu2: {gpu2}");
Console.WriteLine($"gpu1 != gpu2: {gpu1 != gpu2}");

Console.WriteLine("\n--Перегрузка ToString()--");
Console.WriteLine($"{cpu1}");
Console.WriteLine($"{gpu}");

Console.WriteLine("\n--Добавление, удаление, вывод общей стоимости--");
Computer myComputer = new Computer();
myComputer.AddComponent(cpu);
myComputer.AddComponent(ram);
myComputer.AddComponent(gpu);
myComputer.AddComponent(hdd);
myComputer.ShowAllComponents();
Console.WriteLine($"Общая стоимость: {myComputer.GetTotalPrice()} р.");
myComputer.RemoveComponent(hdd);
myComputer.ShowAllComponents();
Console.WriteLine($"Общая стоимость: {myComputer.GetTotalPrice()} р.");

Console.WriteLine("\n--Компьютерная фабрика--");
Console.WriteLine("Офисный компьютер");
Computer officeComputer = ComputerFactory.CreateOfficeComputer();
officeComputer.ShowAllComponents();
Console.WriteLine($"Общая стоимость: {officeComputer.GetTotalPrice()} р.");
Console.WriteLine();
Console.WriteLine("Игровой компьютер");
Computer gameComputer = ComputerFactory.CreateGameComputer();
gameComputer.ShowAllComponents();
Console.WriteLine($"Общая стоимость: {gameComputer.GetTotalPrice()} р.");
Console.WriteLine();
Console.WriteLine("Кастомная сборка");
Computer customComputer = ComputerFactory.CreateCustomComputer(new CPU("Intel N100", "Intel", 5000m, 4, 3.4), new RAM("Crucial SO-DIMM 8GB DDR4-3200", "Crucial", 3000m, 8, 3200));
customComputer.ShowAllComponents();
Console.WriteLine($"Общая стоимость: {customComputer.GetTotalPrice()} р.");
Console.WriteLine();
Console.WriteLine("--Завершение--");
```
