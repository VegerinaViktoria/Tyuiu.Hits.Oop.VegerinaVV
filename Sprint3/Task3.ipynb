<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс BankAccount в C#, который будет представлять
информацию об учетных записях в банке. На основе этого класса разработать 2-3
производных класса, демонстрирующих принципы наследования и полиморфизма.
В каждом из классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.

Требования к базовому классу BankAccount:

• Атрибуты: Номер счета (AccountNumber), Баланс (Balance), Тип счета
(AccountType).

• Методы:

o GetInfo(): метод для получения информации о счете в виде строки.

o Deposit(decimal amount): метод для внесения денег на счет.

o Withdraw(decimal amount): метод для снятия денег со счета.

Требования к производным классам:

1. Сберегательный счет (SavingsAccount): Должен содержать дополнительные
атрибуты, такие как Процентная ставка (InterestRate). Метод Deposit() должен
быть переопределен для добавления процентов к сумме вклада при
внесении денег на счет.

2. Текущий счет (CheckingAccount): Должен содержать дополнительные
атрибуты, такие как Лимит овердрафта (OverdraftLimit).
Метод Withdraw() должен быть переопределен для проверки и применения
лимита овердрафта при снятии денег со счета.

3. Инвестиционный счет (InvestmentAccount) (если требуется третий класс):
Должен содержать дополнительные атрибуты, такие как Список активов
(AssetsList). Метод GetInfo() должен быть переопределен для включения
информации о списках активов в описании счета.

```
using System;
using System.Collections.Generic;

public class BankAccount
{
    private string _accountNumber;
    private decimal _balance;
    private string _accountType;
    private string _ownerName;
    private decimal _dailyLimit;
    private string _email;
    
    public string AccountNumber
    {
        get { return _accountNumber; }
        set { _accountNumber = value; }
    }

    public decimal Balance
    {
        get { return _balance; }
        set { _balance = value; }
    }

    public string AccountType
    {
        get { return _accountType; }
        set { _accountType = value; }
    }

    public string OwnerName
    {
        get { return _ownerName; }
        set { _ownerName = value; }
    }

    public decimal DailyLimit 
    { 
        get { return _dailyLimit; } 
        set { _dailyLimit = value; } 
    } 

    public string Email
    {
        get { return _email; }
        set { _email = value; }
    }

    public BankAccount(string accountNumber, decimal balance, string accountType,
     string ownerName, decimal dailyLimit, string email)
    {
        AccountNumber = accountNumber;
        Balance = balance;
        AccountType = accountType;
        OwnerName = ownerName;
        DailyLimit = dailyLimit;
        Email = email;
    }

    public virtual string GetInfo()
    {
        return $"Номер счета: {AccountNumber}, Тип счета: {AccountType}, Баланс: {Balance} р.";
    }

    public virtual void Deposit(decimal amount)
    {
        Balance = Balance + amount;
        Console.WriteLine($"Внесено на счет {AccountNumber}: {amount} р. Баланс: {Balance} р.");
    }

    public virtual void Withdraw(decimal amount)
    {
        if (Balance >= amount)
        {
            Balance = Balance - amount;
            Console.WriteLine($"Снято со счета {AccountNumber}: {amount} р. Баланс: {Balance} р.");
        }
        else
        {
            Console.WriteLine($"Недостаточно средств, Баланс: {Balance} р.");
        }
    }

    public void ShowOwnerName()
    {
        Console.WriteLine($"Владельцем счета {AccountNumber} является {OwnerName}");
    }

    public virtual void ShowDailyLimit()
    {
        Console.WriteLine($"Дневной лимит снятия денежных средств со счета {AccountNumber}: {DailyLimit} р.");
    }

    public void ShowContactInfo()
    {
        Console.WriteLine($"Контактная информация счета {AccountNumber}: {Email}");
    }
}

public interface IInsurable
{
    void Insure();
}

public interface IRisky
{
    void CheckRiskLevel();
}

public interface IProfitable
{
    void CheckProfit();
}

public class SavingsAccount : BankAccount
{
    private decimal _interestRate;
    private decimal _minBalance;
    private decimal _welcomeRate;
    private int _welcomePeriodMonths;
    private bool _autoRenew;
    private decimal _totalInterestEarned;

    public decimal InterestRate
    {
        get { return _interestRate; }
        set { 
            if (value >= 0 && value <= 100)
                _interestRate = value;
            else
                throw new ArgumentOutOfRangeException("Некорректная процентная ставка!");
            }
    }

    public decimal MinBalance
    {
        get { return _minBalance; }
        set { _minBalance = value; }
    }

    public decimal WelcomeRate
    {
        get { return _welcomeRate; }
        set { 
            if (value >= 0 && value <= 100)
                _welcomeRate = value;
            else
                throw new ArgumentOutOfRangeException("Некорректная приветственная процентная ставка!");
            }
    }

    public int WelcomePeriodMonths
    {
        get { return _welcomePeriodMonths; }
        set { _welcomePeriodMonths = value; }
    }

    public bool AutoRenew
    {
        get { return _autoRenew; }
        set { _autoRenew = value; }
    }

    public decimal TotalInterestEarned
    {
        get { return _totalInterestEarned; }
        set { _totalInterestEarned = value; }
    }

    public SavingsAccount(string accountNumber, decimal balance, string accountType, string ownerName, decimal dailyLimit,
     string email, decimal interestRate, decimal minBalance, decimal welcomeRate, int welcomePeriodMonths, 
     bool autoRenew) : base (accountNumber, balance, accountType, ownerName, dailyLimit, email)
    {
        InterestRate = interestRate;
        MinBalance = minBalance;
        WelcomeRate = welcomeRate;
        WelcomePeriodMonths = welcomePeriodMonths;
        AutoRenew = autoRenew;
    }

    public override void Deposit(decimal amount)
    {
        Balance = Balance + amount;
        decimal percents = decimal.Round(Balance * (InterestRate / 100), 2);
        Balance = Balance + percents;
        TotalInterestEarned += percents;  
        Console.WriteLine($"Внесено на счет {AccountNumber}: {amount} р. Начислены проценты {percents} р. Ставка {InterestRate}%, Баланс: {Balance} р.");
    }

    public void IsGoodInterestRate()
    {
        if (InterestRate >= 5)
        {
            Console.WriteLine($"Ваш вклад выгодный! Процентная ставка: {InterestRate}%");
        }
        else
        {
            Console.WriteLine($"Ваш вклад невыгодный! Низкая процентная ставка: {InterestRate}%.");
        }
    }

    public void ShowMinBalanceInfo()
    {
        Console.WriteLine($"На счете номер {AccountNumber} установлен минимальный баланс: {MinBalance} р.");
    }

    public void CheckWelcomeRate(int monthsPassed)
    {
        if (monthsPassed < WelcomePeriodMonths)
        {
            Console.WriteLine($"На счете {AccountNumber} установлена приветственная ставка: {WelcomeRate}%. Осталось месяцев: {WelcomePeriodMonths - monthsPassed}");
        }
        else
        {
            Console.WriteLine($"На счете {AccountNumber} срок приветственной ставки завершен {WelcomePeriodMonths} мес. Ваша текущая ставка: {InterestRate}%");
        }
    }

    public override void ShowDailyLimit()
    {
        Console.WriteLine($"Дневной лимит снятия денежных средств сберегательного счета {AccountNumber}: {DailyLimit} р.");
    }

    public void ShowAutoRenewStatus()
    {
        if(AutoRenew)
        {
            Console.WriteLine($"Автопродление вклада сберегательного счета {AccountNumber} включено");
        }
        else
        {
            Console.WriteLine($"Автопродление вклада сберегательного счета {AccountNumber} отключено");
        } 
    }

    public void ShowTotalInterest()
    {
        Console.WriteLine($"Всего начислено процентов по счету {AccountNumber}: {TotalInterestEarned} р.");
    }
}

public class PremiumSavingsAccount : SavingsAccount
{
    private int _premiumLevel;
    private decimal _bonusRate;
    private string _bankManagerName;
    private int _supportAnswerTime;

    public int PremiumLevel
    {
        get { return _premiumLevel; }
        set { _premiumLevel = value; }
    }

    public decimal BonusRate
    {
        get { return _bonusRate; }
        set { 
            if (value >= 0 && value <= 100)
                _bonusRate = value;
            else
                throw new ArgumentOutOfRangeException("Некорректная бонусная процентная ставка!");
            }
    }

    public string BankManagerName
    {
        get { return _bankManagerName; }
        set { _bankManagerName = value; }
    }

    public int SupportAnswerTime
    {
        get { return _supportAnswerTime; }
        set
        {
            if (value > 0)
                _supportAnswerTime = value;
            else
                throw new ArgumentOutOfRangeException("Время ответа не может быть отрицательным!");
        }
    }

    public PremiumSavingsAccount(string accountNumber, decimal balance, string accountType, string ownerName, decimal dailyLimit,
    string email, decimal interestRate, decimal minBalance, decimal welcomeRate, int welcomePeriodMonths, int premiumLevel,
    decimal bonusRate, string bankManagerName, bool autoRenew, int supportAnswerTime) : base(accountNumber, balance, accountType, 
    ownerName, dailyLimit, email, interestRate, minBalance, welcomeRate, welcomePeriodMonths, autoRenew)
    {
        PremiumLevel = premiumLevel;
        BonusRate = bonusRate;
        BankManagerName = bankManagerName;
        SupportAnswerTime = supportAnswerTime;
    }

    public void ShowPremiumLevel()
    {
        Console.WriteLine($"Текущий премиальный уровень счета {AccountNumber}: {PremiumLevel}");
    }

    public void ShowBankManagerName()
    {
        Console.WriteLine($"В рамках премиального счета Вам доступен персональный менеджер. Ваш менеджер: {BankManagerName}");
    }

    public override void Deposit(decimal amount)
    {
        base.Deposit(amount);
        decimal bonus = decimal.Round(amount * (BonusRate / 100), 2);
        Balance += bonus;
        Console.WriteLine($"На счет {AccountNumber} начислен премиальный бонус {bonus} р. Баланс: {Balance} р.");
    }

    public override void ShowDailyLimit()
    {
        Console.WriteLine($"Дневной лимит снятия денежных средств премиум-сберегательного счета {AccountNumber}: {DailyLimit} р.");
    }

    public void ShowSupportInfo()
    {
        Console.WriteLine($"Приоритетная поддержка премиум-счета {AccountNumber}: ответ в течение {SupportAnswerTime} мин.");
    }
}

public class CheckingAccount : BankAccount
{
    private decimal _overdraftLimit;
    private decimal _paymentNotifications;
    private decimal _cashbackPercent;
    private bool _hasCard;
    private int _removalsCount;

    public decimal OverdraftLimit
    {
        get { return _overdraftLimit; }
        set { 
            if (value >= 0)
                _overdraftLimit = value; 
            else
                throw new ArgumentOutOfRangeException("Лимит овердрафта не может быть отрицательным!");
            }
    }

    public decimal PaymentNotifications
    {
        get { return _paymentNotifications; }
        set { 
            if (value >= 0)
                _paymentNotifications = value; 
            else
                throw new ArgumentOutOfRangeException("Оплата за уведомления не может быть отрицательной!");
            }
    }

    public decimal CashbackPercent
    {
        get { return _cashbackPercent; }
        set { 
            if (value >= 0)
                _cashbackPercent = value; 
            else
                throw new ArgumentOutOfRangeException("Кэшбек не может быть отрицательным!");
            }
    }

    public bool HasCard
    {
        get { return _hasCard; }
        set { _hasCard = value; }
    }

    public int RemovalsCount
    {
        get { return _removalsCount; }
        set { _removalsCount = value; }
    }

    public CheckingAccount(string accountNumber, decimal balance, string accountType, string ownerName,
     decimal dailyLimit, string email, decimal overdraftLimit, decimal paymentNotifications, 
     decimal cashbackPercent, bool hasCard, int removalsCount) 
     : base(accountNumber, balance, accountType, ownerName, dailyLimit, email)
    {
        OverdraftLimit = overdraftLimit;
        PaymentNotifications = paymentNotifications;
        CashbackPercent = cashbackPercent;
        HasCard = hasCard;
        RemovalsCount = removalsCount;
    }

    public override void Withdraw(decimal amount)
    {
        decimal total = Balance + OverdraftLimit;

        if (total >= amount)
        {
            Balance = Balance - amount;
            RemovalsCount++;
            Console.WriteLine($"Снято со счета {AccountNumber}: {amount} р. Лимит овердрафта: {OverdraftLimit} р. Баланс: {Balance} р.");
        }
        else
        {
            Console.WriteLine($"Превышен лимит овердрафта, Доступно: {total} р.");
        }
    }
    public bool IsOverdraft()
    {
        if (Balance < 0)
        {
            Console.WriteLine($"На вашем счете {AccountNumber} подключена опция овердрафта. Лимит овердрафта: {OverdraftLimit} р. Использовано: {Math.Abs(Balance)} из {OverdraftLimit} р.");
            return true;
        }
        else
        {
            Console.WriteLine($"Счет {AccountNumber} в положительном балансе. Баланс: {Balance} р.");
            return false;
        }
    }

    public void Transfer(BankAccount account, decimal amount)
    {
        if (Balance >= amount)
        {
            Withdraw(amount);
            account.Deposit(amount);
            Console.WriteLine($"Перевод: {amount} р. со счета: {AccountNumber} на счет: {account.AccountNumber}");
        }
        else
        {
            Console.WriteLine($"На счете: {AccountNumber} недостаточно средств для перевода!");
        }
    }

    public void Transfer(string phoneNumber, decimal amount, string info)
    {
        if (Balance >= amount)
        {
            Balance -= amount;
            RemovalsCount++;
            Console.WriteLine($"Перевод со счета {AccountNumber} {amount} р. по СБП на номер {phoneNumber}. Информация: [{info}]. Баланс: {Balance} р.");
        }
        else 
        {
            Console.WriteLine("Недостаточно средств!");
        }
    }

    public void PayNotifications()
    {
        if (Balance >= PaymentNotifications)
        {
            Balance -= PaymentNotifications;
            Console.WriteLine($"Со счета {AccountNumber} списана плата за уведомления: {PaymentNotifications} р. Баланс: {Balance} р.");
        }
        else
        {
            Console.WriteLine($"Недостаточно средств для оплаты уведомлений!");
        }
    }

    public void AccrueCashback(decimal amount)
    {
        decimal cashback = decimal.Round(amount * (CashbackPercent / 100), 2);
        Balance = Balance + cashback;
        Console.WriteLine($"На счет {AccountNumber} начислен кэшбэк: {cashback} р. за операцию: {amount} р. Процент кэшбека: {CashbackPercent}%. Баланс: {Balance} р.");
    }

    public override void ShowDailyLimit()
    {
        Console.WriteLine($"Дневной лимит снятия денежных средств текущего счета {AccountNumber}: {DailyLimit} р.");
    }

    public void ShowCardInfo()
    {
        if(HasCard)
        {
            Console.WriteLine($"К счету {AccountNumber} привязана карта");
        }
        else
        {
            Console.WriteLine($"К счету {AccountNumber} не привязана карта");
        } 
    }

    public void ShowRemovalsCount()
    {
        Console.WriteLine($"Количество транзакций-операций снятия со счета {AccountNumber}: {RemovalsCount}");
    }
}

public class InvestmentAccount : BankAccount, IInsurable, IRisky, IProfitable
{
    private List<string> _assetsList;
    private decimal _riskLevel;
    private decimal _insuranceSum;
    private decimal _investedSum;
    private decimal _current;
    private decimal _dividendYield;

    public decimal RiskLevel
    {
        get { return _riskLevel; }
        set { _riskLevel = value; }
    }

    public decimal InsuranceSum
    {
        get { return _insuranceSum; }
        set { _insuranceSum = value; }
    }

    public decimal InvestedSum
    {
        get { return _investedSum; }
        set { _investedSum = value; }
    }

    public decimal Current
    {
        get { return _current; }
        set { _current = value; }
    }

    public List<string> AssetsList
    {
        get { return _assetsList; }
        set { _assetsList = value; }
    }

    public decimal DividendYield
    {
        get { return _dividendYield; }
        set
        {
            if (value >= 0 && value <= 100)
                _dividendYield = value;
            else
                throw new ArgumentOutOfRangeException("Дивидендная доходность должна быть от 0 до 100!");
        }
    }

    public InvestmentAccount(string accountNumber, decimal balance, string accountType, string ownerName, 
    decimal dailyLimit, string email, List<string> assets, decimal riskLevel, decimal insuranceSum, 
    decimal investedSum, decimal current, decimal dividendYield) 
     : base(accountNumber, balance, accountType, ownerName, dailyLimit, email)
    {
        AssetsList = assets;
        RiskLevel = riskLevel;
        InsuranceSum = insuranceSum;
        InvestedSum = investedSum;
        Current = current;
        DividendYield = dividendYield;
    }

    public void Insure()
    {
        Console.WriteLine($"Счет {AccountNumber} застрахован на {InsuranceSum} р.");
    }

    public void CheckRiskLevel()
    {
        if (RiskLevel <= 3)
        {
            Console.WriteLine("Уровень риска счета: низкий риск");
        }
        else if (RiskLevel <= 7)
        {
            Console.WriteLine("Уровень риска счета: средний риск");
        } 
        else
        {
            Console.WriteLine("Уровень риска  счета: высокий риск");
        }    
    }

    public void CheckProfit()
    {
        decimal profit = Current - InvestedSum;
        Console.WriteLine($"Счет {AccountNumber} принес доход {profit} р.");
    }

    public override string GetInfo()
    {
        string assets = string.Join(", ", AssetsList);
        return $"Номер счета: {AccountNumber}, Тип счета: {AccountType}, Баланс: {Balance} р. , Активы: {assets}";
    }

    public void ShowAssetsCount()
    {
        Console.WriteLine($"Количество активов: {AssetsList.Count}");
    }

    public void ShowDividendYield()
    {
        Console.WriteLine($"Дивидендная доходность счета {AccountNumber}: {DividendYield}%");
    }

    public override void ShowDailyLimit()
    {
        Console.WriteLine($"Дневной лимит снятия денежных средств инвестиционного счета {AccountNumber}: {DailyLimit} р.");
    }

    public void AddAsset(string asset)
    {
        if (AssetsList.Contains(asset))
        {
            Console.WriteLine($"Актив \"{asset}\" уже есть в портфеле счета {AccountNumber}.");
            return;
        }

        AssetsList.Add(asset);
        Console.WriteLine($"Актив {asset} добавлен в портфель счета {AccountNumber}. Всего активов: {AssetsList.Count}");
    }

    public void RemoveAsset(string asset)
    {
        if (!AssetsList.Contains(asset))
        {
            Console.WriteLine($"Актив {asset} не найден в портфеле счета {AccountNumber}.");
            return;
        }

        AssetsList.Remove(asset);
        Console.WriteLine($"Актив {asset} удален из портфеля счета {AccountNumber}. Осталось активов: {AssetsList.Count}");
    }
}

public class BankAccountCollection<T> where T : BankAccount
{
    private List<T> _bankAccounts = new List<T>();

    public void Add(T bankAccount) 
    {
        _bankAccounts.Add(bankAccount);
        Console.WriteLine($"Счет {bankAccount.AccountNumber} добавлен в банковскую систему");
    }
    public void Remove(T bankAccount)
    {
        _bankAccounts.Remove(bankAccount);
        Console.WriteLine($"Счет {bankAccount.AccountNumber} удален из банковской системы");
    }

    public void GetTotalBalance()
    {
        decimal total = 0;
        foreach (var bankAccount in _bankAccounts)
        {
            total += bankAccount.Balance;
        }
        Console.WriteLine($"Общий баланс банка: {total}");
    }

    public void DisplayAllAccounts()
    {
        foreach (var bankAccount in _bankAccounts)
        {
            Console.WriteLine(bankAccount.GetInfo());
        }
    }
}

SavingsAccount savingsAccount = new SavingsAccount("1357", 2000, "Сберегательный", "Виктория Вегерина",15000,"vikavvv@gmail.com",5, 100, 12, 2, true);
PremiumSavingsAccount premiumSavingsAccount = new PremiumSavingsAccount("6574", 150000, "Премиум-сберегательный","Артем Казаков", 100000, "kazakad@mail.ru", 10, 100, 15, 5, 5, 5, "Ульяна Романова", true, 2);
CheckingAccount checkingAccount = new CheckingAccount("2569", 15000, "Текущий", "Богдан Горбунов", 30000, "gorbunbog@yandex.ru",500, 100, 5, true, 0);
List<string> assets = new List<string> {"Акции ВТБ", "Акции СБЕР", "Акции Газпромбанк"};
InvestmentAccount investmentAccount = new InvestmentAccount("4807", 1500000, "Инвестиционный","Полина Ушакова", 50000, "pollyyshakova@mail.ru", assets, 2, 1000, 1000, 50000, 5.5m);

BankAccount bankAccount1 = savingsAccount;
BankAccount bankAccount2 = checkingAccount;
BankAccount bankAccount3 = investmentAccount;
BankAccount bankAccount4 = premiumSavingsAccount;

Console.WriteLine("--Сберегательный счет--");
Console.WriteLine(bankAccount1.GetInfo());
bankAccount1.Deposit(700);
bankAccount1.Withdraw(500);
bankAccount1.ShowOwnerName();
savingsAccount.IsGoodInterestRate();
savingsAccount.ShowMinBalanceInfo();
savingsAccount.CheckWelcomeRate(1);
savingsAccount.ShowDailyLimit();
savingsAccount.ShowContactInfo();
savingsAccount.ShowAutoRenewStatus();
savingsAccount.ShowTotalInterest();
Console.WriteLine();

Console.WriteLine("--Премиум-сберегательный счет--");
Console.WriteLine(bankAccount4.GetInfo());
bankAccount4.Deposit(60000);
bankAccount4.Withdraw(300);
bankAccount4.ShowOwnerName();
premiumSavingsAccount.IsGoodInterestRate();
premiumSavingsAccount.ShowMinBalanceInfo();
premiumSavingsAccount.CheckWelcomeRate(6);
premiumSavingsAccount.ShowPremiumLevel();
premiumSavingsAccount.ShowBankManagerName();
premiumSavingsAccount.ShowDailyLimit();
premiumSavingsAccount.ShowContactInfo();
premiumSavingsAccount.ShowAutoRenewStatus();
premiumSavingsAccount.ShowTotalInterest();
premiumSavingsAccount.ShowSupportInfo();
Console.WriteLine();

Console.WriteLine("--Текущий счет--");
Console.WriteLine(bankAccount2.GetInfo());
bankAccount2.Deposit(200);
bankAccount2.Withdraw(400);
bankAccount2.ShowOwnerName();
checkingAccount.IsOverdraft();
checkingAccount.PayNotifications();
checkingAccount.AccrueCashback(500);
checkingAccount.ShowDailyLimit();
checkingAccount.ShowContactInfo();
checkingAccount.Transfer("89123456789", 1000, "За кофе");
checkingAccount.ShowCardInfo();
checkingAccount.ShowRemovalsCount();
Console.WriteLine();

Console.WriteLine("--Инвестиционный счет--");
Console.WriteLine(bankAccount3.GetInfo());
bankAccount3.Deposit(400);
bankAccount3.Withdraw(1000);
bankAccount3.ShowOwnerName();
investmentAccount.ShowAssetsCount();
investmentAccount.CheckRiskLevel();
investmentAccount.CheckProfit();
investmentAccount.ShowDailyLimit();
investmentAccount.ShowContactInfo();
investmentAccount.ShowDividendYield();
investmentAccount.AddAsset("Акции Т-банк");
investmentAccount.RemoveAsset("Акции Газпромбанк");
Console.WriteLine();

// Взаимодействия объектов
Console.WriteLine("--Переводы--");
checkingAccount.Transfer(savingsAccount, 700);
checkingAccount.Transfer(investmentAccount, 1500);
Console.WriteLine();
//Generic-класс
BankAccountCollection<BankAccount> accounts = new BankAccountCollection<BankAccount>();
Console.WriteLine("--Добавление--");
accounts.Add(savingsAccount);
accounts.Add(checkingAccount);
accounts.Add(investmentAccount);
accounts.Add(premiumSavingsAccount);
Console.WriteLine("--Счета банка--");
accounts.DisplayAllAccounts();
accounts.GetTotalBalance();
```
