```polyglot-notebook
public class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; }

    public Plant(string name, string type, float height)
    { 
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Bloom()
    {
        Console.WriteLine($"Растение {Name} цветет");
    }

    public void Grow(int growth)
    {
        Height += growth;
        Console.WriteLine($"Растение {Name} выросло на {growth} см. Текущая высота: {Height} см");
    }
}

public class Flower : Plant
{
    public string Color { get; set; }
    public int NumPetals { get; set; }

    public Flower(string name, float height, string color, int numPetals) : base(name, "Цветок", height)
    {
        Color = color;
        NumPetals = numPetals;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Цветок {Name} распускается и радует глаз!");
    }

    public void Wither()
    {
        Console.WriteLine($"Цветок {Name} начинает увядать");
    }
}

public class Tree : Plant
{
    public int Age { get; set; }
    public string TypeLeaves { get; set; }

    public Tree(string name, float height, int age, string typeLeaves) : base(name, "Дерево", height)
    {
        Age = age;
        TypeLeaves = typeLeaves;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Дерево {Name} покрывается новыми листьями!");
    }

    public void GrowOlder(int years)
    {
        Age += years;
        Console.WriteLine($"Дерево {Name} стало старше на {years} лет. Текущий возраст: {Age} лет");
    }
}

Flower myFlower = new Flower("Пион", 20, "Красный", 5);
Tree myTree = new Tree("Ель", 200, 8, "Хвойные");

Console.WriteLine("--Цветок--");
Console.WriteLine($"Растение: {myFlower.Name}, Тип: {myFlower.Type}, Высота: {myFlower.Height} см, Цвет: {myFlower.Color}, Количество лепестков: {myFlower.NumPetals}");
myFlower.Bloom();
myFlower.Grow(25);
myFlower.Wither();

Console.WriteLine("--Дерево--");
Console.WriteLine($"Растение: {myTree.Name}, Тип: {myTree.Type}, Высота: {myTree.Height} см, Возраст: {myTree.Age} лет, Тип листьев: {myTree.TypeLeaves}");
myTree.Bloom();
myTree.Grow(50);
myTree.GrowOlder(3);
```
