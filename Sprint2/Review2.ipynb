```polyglot-notebook
public class Employee
{
    private string _name;
    private string _position;
    private decimal _salary;

   public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public string Position
    {
        get { return _position; }
        set { _position = value; }
    }

    public decimal Salary
    {
        get { return _salary; }
        set { _salary = value; }
    }

    public Employee(string name, string position, decimal salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    public virtual void PrintInfo()
    {
        Console.WriteLine($"Имя сотрудника: {Name}, Должность: {Position}, Зарплата: {Salary} р.");
    }
}

public class Manager : Employee
{
    private string _department;

    public string Department
    {
        get { return _department; }
        set { _department = value; }
    }

    public Manager(string name, string position, decimal salary, string department) : base(name, position, salary)
    {
        Department = department;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Отдел: {Department}");
    }
}

public interface IReportable
{
    string GenerateReport();
}

public interface ITeamLeader
{
    string[] GetTeamMembers();
}

public class ProjectManager : Manager, IReportable, ITeamLeader
{
    private string _projectName;

    public string ProjectName
    {
        get { return _projectName; }
        set { _projectName = value; }
    }
    
    private string[] _teamMembers;

    public ProjectManager(string name, string position, decimal salary, string department, string projectName,
     string[] teamMembers) : base(name, position, salary, department)
    {
        ProjectName = projectName;
        _teamMembers = teamMembers;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Название проекта: {ProjectName}");
    }

    public string GenerateReport()
    {
        return $"Проект: {ProjectName}, Проектный менеджер: {Name}, Отдел: {Department}, Статус: Все задачи выполнены.";
    }

    public string[] GetTeamMembers()
    {
        return _teamMembers;
    }
}

Employee employee = new Employee("Вероника", "Веб-дизайнер", 70000);
Manager manager = new Manager("Полина", "Менеджер", 80000, "Отдел продаж");
ProjectManager projectManager = new ProjectManager("Виктория", "Проектный менеджер", 100000, "Проектный отдел", "Разработка веб-сайта ITHighSchoolSpace", new string[]{"Вероника", "Полина", "Виктория"});
Console.WriteLine("--Сотрудник--");
employee.PrintInfo();
Console.WriteLine();
Console.WriteLine("--Менеджер--");
manager.PrintInfo();
Console.WriteLine();
Console.WriteLine("--Проектный менеджер--");
projectManager.PrintInfo();
Console.WriteLine(projectManager.GenerateReport());
Console.WriteLine();
Console.WriteLine($"Команда проекта '{projectManager.ProjectName}':");
foreach (string member in projectManager.GetTeamMembers())
{
    Console.WriteLine($"~ {member}");
}
```
