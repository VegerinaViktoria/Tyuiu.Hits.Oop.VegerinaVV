```polyglot-notebook
public class Product
{
    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public decimal Price
    {
        get { return price; }
        set { price = value; }
    }

    public int Quantity
    {
        get { return quantity; }
        set { quantity = value; }
    }

    public string Category
    {
        get { return category; }
        set { category = value; }
    }

    public string Description
    {
        get { return description; }
        set { description = value; }
    }

    public Product(string name, decimal price, int quantity, string category, string description)
    {
        Name = name;
        Price = price;
        Quantity = quantity;
        Category = category;
        Description = description;
    }

    public decimal GetTotalPrice()
    {
        return Quantity * Price;
    }

    public void UpdateQuantity(int amount)
    {
        Quantity += amount;
        Console.WriteLine($"Количество товара {Name} обновлено на {amount} шт. Текущее количество товара: {Quantity} шт.");
    }

    public string GetProductInfo()
    {
        return $"Товар: {Name}, Категория: {Category}, Количество: {Quantity} шт, Цена: {Price} р, Описание: {Description}";
    }
}

public class Seller
{
    private string name;
    private string employeeId;
    protected decimal salary;
    internal string contactInfo;
    private List<Product> productsList;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public string EmployeeId
    {
        get { return employeeId; }
        set { employeeId = value; }
    }

    public decimal Salary 
    {
        get { return salary; }
        set { salary = value; }
    }

    public string ContactInfo
    {
        get { return contactInfo; }
        set { contactInfo = value; }
    }

    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        Name = name;
        EmployeeId = employeeId;
        Salary = salary;
        ContactInfo = contactInfo;
        productsList = new List<Product>();
    }

    public void AddProduct(Product product)
    {
        productsList.Add(product);
        Console.WriteLine($"Товар {product.Name} добавлен в список товаров продавца {Name}");
    }

    public void SellProduct(Product product, int quantity)
    {
        if (product.Quantity >= quantity)
        {
            product.UpdateQuantity(-quantity);
        }
        else
        {
            Console.WriteLine($"Недостаточно товара {product.Name} на складе!");
        }
    }

    public string GetSellerInfo()
    {
        return $"Продавец: {Name}, ID сотрудника: {EmployeeId}, Зарплата: {Salary} р, Контактная информация: {ContactInfo}";
    }

    public List<Product> GetProducts()
    {
        return productsList;
    }
}

public class Store
{
    private string storeName;
    private string location;
    public string storeHours;
    private List<Seller> sellersList;

    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    public string Location
    {
        get { return location; }
        set { location = value; }
    }

    public Store(string storeName, string location, string storeHours)
    {
        StoreName = storeName;
        Location = location;
        this.storeHours = storeHours;
        sellersList = new List<Seller>();
    }

    public void AddSeller(Seller seller)
    {
        sellersList.Add(seller);
        Console.WriteLine($"Продавец {seller.Name} добавлен в магазин {StoreName}");
    }

    public void ListProducts()
    {
       foreach (Seller seller in sellersList)
       {

        foreach (Product product in seller.GetProducts())
        {
            Console.WriteLine(product.GetProductInfo());
        }

       }
    }

    public string GetStoreInfo()
    {
        return $"Название магазина: {StoreName}, Адрес: {Location}, Часы работы: {storeHours}";
    }
}

Product product = new Product("Смартфон", 50000, 10, "Электроника", "Realme GT 7");
Seller seller = new Seller("Виктория", "1234", 70000, "viktoriav@gmail.com");
Store store = new Store("М.Видео","г. Тюмень ул. Широтная 80", "10:00 - 22:00");

Console.WriteLine(product.GetProductInfo());
product.UpdateQuantity(2);
Console.WriteLine($"Текущая цена за {product.Quantity} смартфонов: {product.GetTotalPrice()}");

Console.WriteLine(seller.GetSellerInfo());
seller.AddProduct(product);
seller.SellProduct(product, 5);

Console.WriteLine(store.GetStoreInfo());
store.AddSeller(seller);
Console.WriteLine("Товаров в магазине: ");
store.ListProducts()
```
